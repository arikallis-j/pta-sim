```sage
phi, gamma = var('phi gamma', domain='real')
a, b, c, d = var('A B C D', domain='real')
# A, B, C, D = a, b+1, c, d-1
# assume(A>0, A<1, C>0, C<1, B>0, B<2, D>-2, D<0)
# J = sin(phi)**2 * (C * cos(phi - gamma) + D) / (A*cos(phi) + B)
assume(A>0, A<1, C>0, C<1, B>0, B<2, D>0, D<2)
H = sin(phi)*sin(phi-gamma) / ((A*cos(phi) + B) * (C*cos(phi-gamma) + D))
f_phi = H
f_phi = f_phi.simplify_full()
f_phi
```

```sage
assumptions()
```

```sage
def to_contour_integral(expr, phi, z):
    """
    expr : подынтегральное выражение f(phi) dphi
    возвращает подынтегральную функцию g(z) для контура |z|=1
    """
    subs = {cos(phi): (z + 1/z)/2, sin(phi): (z - 1/z)/(2*I)}  # если нужен синус
    expr_z = expr.subs(subs) * (1/(I*z))
    return expr_z.full_simplify()

z = var('z')
f_z = to_contour_integral(f_phi, phi, z)
f = f_z
f
```

```sage
g = f.denominator()
g
```

```sage
roots = solve(g == 0, z)
S, Q = var('S Q')
z_roots = [r.rhs() for r in roots]
z_roots
```

```sage
z_roots_contour = [ r for r in z_roots if limit(r, A=0) == 0 or limit(r, C=0) == 0]
# z_roots_contour = [z.subs({sqrt(B^2 - A^2): S, sqrt(D^2-C^2): Q}).full_simplify() for z in z_roots_contour]
z_roots_contour
```

```sage
residues = [f.maxima_methods().residue(z, z0) for z0 in z_roots_contour]
res_sum = 0
for r in residues:
    res_sum += r
(res_sum*I).simplify_rational()
```

```sage
Int_f = (I * res_sum).full_simplify()
Int_f = Int_f.factor().subs({sqrt(B^2 - A^2): S}).subs({A^2: B^2 - S^2, A^4: (B^2 - S^2)^2, A^6:(B^2 - S^2)^3, A^8:(B^2-S^2)^4}).subs({B-S: B-S}).factor()
Int_f = Int_f.factor().subs({sqrt(B^2 - A^2): S}).subs({A^3: (B^2 - S^2)*A, A^5: (B^2 - S^2)^2*A, A^7:(B^2 - S^2)^3*A, A^9:(B^2-S^2)^4*A}).subs({B-S: B-S}).factor()
Int_f
```
